# FAISS local evidence retrieval

This notebook loads the persisted index through its recorded embedding contract and filters out passages with no meaningful term overlap.


## Setup

Import dependencies and define the retrieval data location.

```python
"""Retrieve evidence from the validated local VERIMIND FAISS index."""

from pathlib import Path

from verimind.faiss_retrieval import FaissEvidenceRetriever

ROOT = Path(__file__).resolve().parent
```

## Validated index loading

The index manifest identifies its embedding backend, dimension, normalization, and metric. Loading fails explicitly if these do not match.


```python
RETRIEVER = FaissEvidenceRetriever.from_index_dir(ROOT / "data" / "index")
```

## Retrieval

Results must share informative terms with the query; this prevents hash-vector collisions from surfacing unrelated pages.


```python
INDEX = RETRIEVER.index
METADATA = RETRIEVER.passages
MODEL = RETRIEVER.model
```

## Search evidence

Queries use the same encoder and normalization contract used to build the loaded index.


```python
def retrieve_evidence(claim: str, k: int = 5) -> list[dict]:
    return RETRIEVER.search(claim, k=k)
```
